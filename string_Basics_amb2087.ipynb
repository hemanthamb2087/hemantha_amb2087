```python
t="Today is 19-Jan-2026 MONDAY"
v=['a','e','i','o','u','A','E','I','O','U']
Count=0
for i in zip(t,v):
  if i ==v:
    Count+=1
print(Count)
```

```python
t="Today is 19-Jan-2026 MONDAY"
v=c=n=o=s=0
for i in t:
  if i.isdigit():
    n+=1
  elif i.isspace():
    s+=1
  elif i in'aeiouAEIOU':
    v+=1
  elif i.isalpha():
    c+=1
  else:
    o+=1
print("The number of digits in the text provided is",n)
print("The number of spaces in the text provided is",s)
print("The number of vowels in the text provided is",v)
print("The number of consonants in the text provided is",c)
print("The number of other characters in the text provided is",o)
```
