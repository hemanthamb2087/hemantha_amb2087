```python
a=float(input("Enter the pH of your soil "))
if(a<5.5):
  print("Your soil is acidic as {} is less than 5.5". format(a))
elif(a<=7.5):
   print("Your soil is neutral as {} is less than 7.5". format(a))
elif(a>=8.5):
   print("Your soil is alkaline as {} is more than 8.5". format(a))
else:
   print("Your soil is highly alkaline")

```
