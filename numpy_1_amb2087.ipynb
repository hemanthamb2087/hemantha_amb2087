```python
import numpy as np
a=np.array([1,2,3,4,5])
print(a)
print(type(a))           #typoe of dataset
print(a.dtype)           #type of datatype of elements
print(a.shape)           #Length of the array along the dimension.Returns as tuple
print(a.size)            #size in terms of number of elements
print(a.itemsize)        #size in bytes
print(a.ndim)            #dimension of the array
print(a.nbytes)          #total size in bytes
```

```python
import numpy as np
arr1=np.array([[1,2],[3,4]])
arr2=np.array([[5,6],[7,8]])
print(arr1)
print(arr2)
print(arr1+arr2)
print(arr1-arr2)
print(arr1*arr2)
print(arr1/arr2)
print(arr1@arr2)
```

```python
concatenated_arr1=np.concatenate((arr1,arr2),axis=1)
print(concatenated_arr1)
```

```python
concatenated_arr1=np.concatenate((arr1,arr2),axis=0)
print(concatenated_arr1)
```

```python
transposed_arr1=arr1.transpose()
print(arr1)
print(transposed_arr1)
```

```python
flattened_arr1=arr1.flatten()
print(arr1)
print(flattened_arr1)
```

```python
import numpy as np
a=np.array([1+1j,1+0j,4.5,3,2,2j])
print("Original array")      #Everything is converted inrto float
print(a)
print("Checking for complex number")   #a=i+p format for complex number
print(np.iscomplex(a))
print("Checking for real number")
print(np.isreal(a))
print("Checking for scalar type")
print(np.isscalar(3.1))
print(np.isscalar([3.1]))
```
