```python
a={1,2,3,4,5,6}
b={4,6,7,8,9,1}
avg_1=sum(a)/len(a)
avg_2=sum(b)/len(b)
if avg_1>avg_2:
  print("The average of all the elements of the set {} is greater than set {}". format(a,b))
else:
  print("The average of all the elements of the set {} is greater than set {}". format(b,a))
```
