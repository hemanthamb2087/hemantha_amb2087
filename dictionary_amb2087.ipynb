```python
keys=["movie","hero","year"]
values=["rajsaab","prabhas",2026]
dic={}
for i in range(len(keys)):
    dic[keys[i]]=values[i]
print(dic)
```

```python
keys=["movie","hero","year"]
values=["rajsaab","prabhas",2026]
dic={}
for i in zip(keys,values):
    dic[i[0]]=i[1]
print(dic)
```
