```python
codes=[                              #extracting and cleaning in one go
  "CRP_RICE_2026",
"CRP_WHEAT_2025",
"FRT_MANGO_2026",
"CRP_MAIZE_2024"
]
clean_codes=[]
for code in codes:
  if code.startswith("CRP_"):
    clean_codes.append(code[4:])
print(clean_codes)
```

```python
split_codes=[]
for code in codes:
  if code.startswith("CRP_"):
    split_codes.append(code.split("_"))
print(split_codes)
```
