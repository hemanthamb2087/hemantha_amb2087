```python
import pandas as pd
data={
    "Farm":["Farm A","Farm B","Farm C","Farm D"],
    "Crop":["Wheat","Rice","Wheat","Maize"],
    "Area_in_hectares":[10,15,8,12],
    "Production_in_tons":[30,60,24,36]
}
df=pd.DataFrame(data,index=[1,2,3,4])
print(df)
```

```python
df.head(2)
```

```python
df.tail(2)
```

```python
df.info()
```

```python
df.describe()
```

```python
df["Yield_ton_per_hectare"]=df["Production_in_tons"]/df["Area_in_hectares"]
print(df)
```

```python
print(df.loc[[1,3]])
```

```python
print(df.loc[[1,3],[ "Farm", "Crop","Area_in_hectares" ,"Production_in_tons"]])
```

```python
print(df["Crop"]=="Wheat")
print()
wheat_farms=df[df["Crop"]=="Wheat"]
print(wheat_farms)
```

```python
high_yield=df[df["Production_in_tons"]>35]
print(high_yield)
```

```python
sorted_df=df.sort_values(by="Production_in_tons",ascending=False)
print(sorted_df)
```

```python
avg_yield=df.groupby("Crop")["Yield_ton_per_hectare"].mean()
print(avg_yield)
```
