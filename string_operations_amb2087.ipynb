```python
crops=["Rice","Wheat","Maize","Millet","Pulses","Cotton"]
print(crops[:3])     #first 3 crops
print(crops[1:5])    #middle crops, excluding end crops
print(crops[-5:-1])   #middle crops, excluding end crops
print(crops[-2:])     #last 2 crops
print(crops[::-1])
print(crops[1:-1])    #remove first and last crop
```

```python
x=[10,20,30,40,50,60]
print(x[1:6])
print(x[1:])
print(x[1:6:2])    #listname[start,end,step]
print(x[1:6:3])
print(x[2:4:1])
print(x[4:2:1])
print(x[::-1])     #reversing the list
print(x[-5:-2:-1])
print(x[-2:-5:-1])
print(x[-2:-5:1])
print(x[-3:-5:-1])
```

```python
#create new list by each element incremented by 2
numbers=[10,20,30,40,50,60]
n=[]
for i in numbers:
  n.append(i+2)
print("New list is",n)
print("Original list is",numbers)
```

```python
#list comprehension create new list by each element incremented by 2
numbers=[10,20,30,40,50,60]
n=[i+2 for i in numbers]
print("New list is",n)
print("Original list is",numbers)
```

```python
numbers=[10,20,30,40,50,60]
n=[]
for i in numbers:
  if i%4==0:
    n.append(i)
print("Numbers divided by 4",n)
```

```python
numbers=[10,20,30,40,50,60]
n=[i for i in numbers if i%4==0]
print("Numbers divisible by 4",n)
```

```python
numbers=[10,21,30,41,50,61]
n=["Even" if i%2==0 else "Odd"for i in numbers]
print(n)

```
