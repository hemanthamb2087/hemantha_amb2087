```python
import random
inventory={}
total_money=0
fish_prices={
    "Rohu":12,
    "Catla":18,
    "Mrigal":25
}
def show_menu():
 print("\n🎣 --- Fishing Game Menu --- 🎣")
 print("1️⃣ Catch Fish")
 print("2️⃣ View Inventory")
 print("3️⃣ Sell Fish")
 print("4️⃣ Exit 🚪")

def catch_fish():
  fish=random.choice(list(fish_prices.keys()))
  inventory[fish]=inventory.get(fish,0)+1
  print(f"You caught a {fish}!!")

def view_inventory():
  if inventory=={}:
    print("No fish")
  else:
    print("Your inventory")
    for fish,count in inventory.items():
      print(f"{fish}:{count}")

def sell_fish():
  global total_money
  if inventory=={}:
    print("No fish to sell")
  else:
    current_sale_earnings = 0
    for fish,count in list(inventory.items()): # Iterate over a copy of items
      current_sale_earnings += count * fish_prices[fish]
      total_money += current_sale_earnings
      inventory.clear()
      print(f"You sold all the fish for ${current_sale_earnings}")

def exit_game():
  print("Thanks for playing!")
  print(f"Total money earned: ${total_money}")
  exit()

show_menu()
while True:
  choice=int(input("Enter your choice: "))

  if choice==1:
    catch_fish()

  elif choice==2:
    view_inventory()

  elif choice==3:
    sell_fish()

  elif choice==4:
    exit_game()
    break
  else:
    print("Invalid choice. Please try again.")
```
