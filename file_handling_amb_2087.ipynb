```python
import random
inventory = {}
total_money = 0
fish_prices = {
   "Rohu": 12,
    "Catla": 18,
    "Sinchu": 25
}
def showmenu():
   print("\n🎣 --- Fishing Game Menu --- 🎣")
   print("1️⃣ Catch Fish")
   print("2️⃣ View Inventory")
   print("3️⃣ Sell Fish")
   print("4️⃣ Exit 🚪")

def catchfish():
        fish = random.choice(list(fish_prices.keys()))
        inventory[fish] = inventory.get(fish, 0) + 1
        print(f"🎉 You caught a {fish}!")
def viewinventory():
        if inventory == {}:
            print("❌ No fish.")
        else:
            print("📦 Your Inventory:")
            for fish, count in inventory.items():
                print(f"{fish}: {count}")
def sellfish():
        global total_money
        if inventory == {}:
            print("❌ No fish to sell.")
        else:
            earnings = 0
            for fish, count in inventory.items():
                earnings += fish_prices[fish] * count
            total_money += earnings
            print(f"💰 You sold all fish for ${earnings}!")
def exitgame():
        print(f"🏁 Final money earned: ${total_money}")
        print("👋 Thanks for playing!")
        exit()


showmenu()
while True:
    choice = int(input("Enter your choice: "))

    if choice == 1:
        catchfish()
    elif choice == 2:
        viewinventory()
    elif choice == 3:
        sellfish()
    elif choice == 4:
        exitgame()
        break
    else:
        print("Invalid choice. Please try again.")
```
