```python
cgpa={1:7.8,2:9.1,3:8.3}
print(cgpa)
print(type(cgpa))
print(len(cgpa))
print("CGPA of 1 is ",cgpa[1])
print("CGPA of 2 is ",cgpa[2])
print("CGPA of 3 is ",cgpa[3])
for k in cgpa:
  print("CGPA of",k,"is\n",cgpa[k])
print("Keys",cgpa.keys())
print("Values",cgpa.values())
print("Items",cgpa.items())
cgpa[4]=6.5
print(cgpa)
cgpa[5]=6.5
print(cgpa)
cgpa[5]=3.5
print(cgpa)
s=0
for i in cgpa:
  s=s+cgpa[i]
print("the total cgpa is",s)
print("the average cgpa is",s/len(cgpa))
print(max(cgpa.values()))
print(min(cgpa.values()))
```
