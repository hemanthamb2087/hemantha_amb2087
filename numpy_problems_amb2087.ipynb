```python
import numpy as np
crop_yields=np.array([
    [12.5,13.2,14.1,13.8,15.2],
    [10.3,11.5,13.8,16.8,23.1],
    [15.1,14.8,16.2,15.5,16.8],
    [11.2,10.9,12.4,11.7,12.8]
])
print("Crop yield array(tonnes):\n",crop_yields)
print(f"The shape of the array is {crop_yields.shape}")
print(f"The size of array is {crop_yields.size}")
print(f"The dimension of array is {crop_yields.ndim}")
print(f"The datatype of the array is {crop_yields.dtype}")
print(f"Total yield:",np.sum(crop_yields))
```

```python
avg_yield_per_farm=np.mean(crop_yields,axis=1)
print("Average yield per farm:",avg_yield_per_farm)
avg_yield_per_season=np.mean(crop_yields,axis=0)
print("Average yield per season:",avg_yield_per_season)
print("The maximum yield is",np.max(crop_yields))
print("The minimum yield is",np.min(crop_yields))
best_farm=np.argmax(avg_yield_per_farm)
print("The best farm is: Farm",best_farm+1)
worst_farm=np.argmin(avg_yield_per_farm)
print("The worst farm is: Farm",worst_farm+1)
best_season=np.argmax(avg_yield_per_season)
print("The best season is: Season",best_season+1)
worst_season=np.argmin(avg_yield_per_season)
print("The worst season is: Season",worst_season+1)
median_yield_per_farm=np.median(crop_yields,axis=1)
print("Median yield per farm:",median_yield_per_farm)
median_yield_per_season=np.median(crop_yields,axis=0)
print("Median yield per season:",median_yield_per_season)
std_yield_per_farm=np.std(crop_yields,axis=1)
print("Standard deviation of yield per farm:",std_yield_per_farm)
std_yield_per_season=np.std(crop_yields,axis=0)
print("Standard deviation of yield per season:",std_yield_per_season)
var_yield_per_farm=np.var(crop_yields,axis=1)
print("Variance of yield per farm:",var_yield_per_farm)
var_yield_per_season=np.var(crop_yields,axis=0)
print("Variance of yield per season:",var_yield_per_season)
```

```python
exceptional=crop_yields >16
print(exceptional)
print("Exceptional yields > 16 are:")
print(crop_yields[exceptional])
```
