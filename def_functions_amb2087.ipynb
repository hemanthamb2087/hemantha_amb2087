```python
import tkinter as tk
root=tk.Tk()
root.geometry("400*400")
label=tk.label(root,text="Hello, Tkinter1",font=("Arial",16))
label.pack(padx=20,pady=20)
def change_text():
  label.config(text="Button clicked")
button=tk.button(root,text="Click me",common=change_text)
button.pack()
root.mainloop()
```

```python
#scope of variables
def func(x):
  print("Local x is",x)
  x=2
  print("Changed local x to",x)
x=50
func(x)
print("x is still",x)
```

```python
#local variable in function
def func1():
  y=2
  print("Changed y to",y)
y=50
func1()
print("Value of y is still",y)
```

```python
#global variable in functions
def func2():
  global z
  z=2
  print("Changed z to",z)
z=50
func2()
print("Value of z is",z)
```

```python
total=0                     #function definition is here
def sum(arg1,arg2):
  total=arg1+arg2           #add both parameters and return them
  print("Inside the function local total:",total)        #total is the local variable
  return total
sum(10,20)                  #call the sum function
print("Outside the function global total:",total)
```

```python
#types of functions
#Required argument
def printme(str):
  print(str)
  return
#printme()   gives an error
printme("Hello")
printme(1234)
```

```python
#Keyword argument/named argument
def printme(str):
  print(str)
  return
printme("Hello")
printme(str="My string")
```

```python
def func(a,b=5,c=10):
  print("a =",a,",b =",b,",c =",c)
func(3,7)   # a=3->required   b=7->keyword   c=10-.default
func(25,c=24) #a=25->required   b=24->default   c=24->keyword
func(c=50,a=100)  #a=100->keyword b=5->default c=50->keyword
```
