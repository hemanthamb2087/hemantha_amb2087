```python
keys=["movie","hero","release year"]
values=["Kantara","Rishab",2022]
mn={}
for i in range(len(keys)):
   mn[keys[i]]=values[i]
print(mn)
```

```python
keys=["movie","hero","release year"]
values=["Kantara","Rishab",2022]
mn={}
for i in zip(keys,values):
   mn[i[0]]=i[1]
print(mn)
```
